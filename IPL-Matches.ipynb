```python
iplDatadf['Venue']=iplDatadf['Venue'].str.split(',').str.get(0)
```

```python
cat_col=['City','MatchNumber','Team1','Team2','Venue',
         'TossWinner','TossDecision','SuperOver','WinningTeam',
         'WonBy','Player_of_Match','Umpire1','Umpire2']
iplDatadf[cat_col]=iplDatadf[cat_col].astype('category')
```

```python
iplDatadf['Date']=pd.to_datetime(iplDatadf['Date'])
```

```python
team_mapping={
    'Delhi Daredevils':'Delhi Capitals',
    'Kings XI Punjab': 'Punjab Kings',
    'Rising Pune Supergiant':'Rising Pune Supergiants'
    }

iplDatadf['Team1']=iplDatadf['Team1'].replace(team_mapping)
iplDatadf['Team2']=iplDatadf['Team2'].replace(team_mapping)
iplDatadf['WinningTeam']=iplDatadf['WinningTeam'].replace(team_mapping)
iplDatadf['TossWinner']=iplDatadf['TossWinner'].replace(team_mapping)
```

### Player_of_Match & Season

```python
def Decision(group):
    return group['Player_of_Match'].value_counts().idxmax()

top_player=(iplDatadf.groupby('Season').apply(Decision))

data=pd.crosstab(iplDatadf['Season'],iplDatadf['Player_of_Match'])[top_player]

plt.figure(figsize=(12,6))

sns.heatmap(data,annot=True,cmap='YlOrRd')
```
